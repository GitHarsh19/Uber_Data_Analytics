```python
df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].fillna('Unknown')
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].fillna('Unknown')
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].fillna('Unknown')
df['Payment Method'] = df['Payment Method'].fillna('Unknown')
df['Avg VTAT'] = df['Avg VTAT'].fillna(0)
df['Avg CTAT'] = df['Avg CTAT'].fillna(0)
df['Cancelled Rides by Customer'] = df['Cancelled Rides by Customer'].fillna(0)
df['Cancelled Rides by Driver'] = df['Cancelled Rides by Driver'].fillna(0)
df['Incomplete Rides'] = df['Incomplete Rides'].fillna(0)
df['Booking Value'] = df['Booking Value'].fillna(0)
df['Ride Distance'] = df['Ride Distance'].fillna(0)
df['Driver Ratings'] = df['Driver Ratings'].fillna(0)
df['Customer Rating'] = df['Customer Rating'].fillna(0)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Booking ID'] = df['Booking ID'].astype('string')
df['Booking Status'] = df['Booking Status'].astype('string')
df['Customer ID'] = df['Customer ID'].astype('string')
df['Vehicle Type'] = df['Vehicle Type'].astype('string')
df['Pickup Location'] = df['Pickup Location'].astype('string')
df['Drop Location'] = df['Drop Location'].astype('string')
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].astype('string')
df['Cancelled Rides by Customer'] = df['Cancelled Rides by Customer'].astype('int')
df['Reason for cancelling by Customer'] = df['Reason for cancelling by Customer'].astype('string')
df['Cancelled Rides by Driver'] = df['Cancelled Rides by Driver'].astype('int')
df['Driver Cancellation Reason'] = df['Driver Cancellation Reason'].astype('string')
df['Incomplete Rides'] = df['Incomplete Rides'].astype('int')
df['Incomplete Rides Reason'] = df['Incomplete Rides Reason'].astype('string')
df['Booking Value'] = df['Booking Value'].astype('int')
df['Payment Method'] = df['Payment Method'].astype('string')
```

```python
total_rides = len(df) # number of rides
print(f"Number of Rides: {total_rides}")
rides_completed = (df['Booking Status'] == 'Completed').sum()
print(f"Number of Rides Completed: {rides_completed}")
```

```python
sucess_rate_rides = (rides_completed/total_rides)*100
print(f"Sucess Rate: {sucess_rate_rides : .2f}%")
```

```python
avg_wait_time = np.mean(df['Avg VTAT'])
print(f"Average Wait Time: {avg_wait_time : .2f}")
```

```python
completed_rides = df[df['Booking Status'] == 'Completed']
total_trip_duration = completed_rides['Avg CTAT'].sum()
avg_trip_duration = total_trip_duration / completed_rides.shape[0]
print(f"Average Trip Duration: {avg_trip_duration: .2f} minutes")
```

```python
completed_rides = df[df['Booking Status'] == 'Completed'] # Filtering just to the Completed Rides
total_ride_distance = completed_rides['Ride Distance'].sum() # Sum all's ride distance
avg_ride_distance = total_ride_distance / completed_rides.shape[0]
print(f"Average Ride Distance: {avg_ride_distance: .2f} minutes")
```

```python
# Calculating Cancellation Rate
rides_cancelled = (
    df['Cancelled Rides by Customer'].sum() +
    df['Cancelled Rides by Driver'].sum()
)  # number of cancelled rides
print(f"Number of Rides Cancelled: {rides_cancelled}")
cancellation_rate = (rides_cancelled/total_rides) * 100
print(f"Cancellation Rate: {cancellation_rate : .2f}%")
```

```python
print(f"Cancellation Rate by Customer: {(df['Cancelled Rides by Customer'].sum()/total_rides)*100 : .2f}%")
print(f"Cancellation Rate by Driver: {(df['Cancelled Rides by Driver'].sum()/total_rides)*100 : .2f}%")
```

```python
df['First Usage'] = df.groupby('Customer ID')['Date'].transform('min') # Calculate First Use Date for each User
df['Last Usage'] = df.groupby('Customer ID')['Date'].transform('max') # Calculate Last Use Date for each User
df['Events After First Usage'] = df.groupby('Customer ID')['Date'].transform('count') # Identifying for each user if have more than 1 event

retained_users = df[df['Events After First Usage'] > 1]['Customer ID'].nunique() # Identifying retained users
first_users = df['Customer ID'].nunique() # Identifying the unique users

retention_rate = (retained_users / first_users) * 100 # Calculating Retention Rate
print(f"Retention Rate: {retention_rate: .2f}%")
```

```python
# Calculating Monthly Active Users
df['Month'] = df['Date'].dt.to_period('M')
mau = df.groupby('Month')['Customer ID'].nunique()

print(f"Monthly Active Users - 2024 {mau}")
```

```python
# Calculate Usage Frequency per Month
month_rides = df.groupby(['Month'])['Booking ID'].count().reset_index(name='Ride Count') # Calculating rides per month
month_usage_frequency = month_rides.groupby('Month')['Ride Count'].mean()

print(f"Usage Frequency per Month {month_usage_frequency}")
```

```python
# Calculate Average Revenue Per Ride
revenue_ride = df['Booking Value'].sum() / total_rides
print(f"Total Revenue: ${df['Booking Value'].sum():.2f}")
print(f"Total Revenue per Ride: ${revenue_ride:.2f}")
```

```python
# Calculate Average Revenue Per User
arpu = df['Booking Value'].sum() / df['Customer ID'].nunique()
print(f"ARPU (Average Revenue Per User): ${arpu:.2f}")
```

```python
# Calculate Driver Average Ratings
avg_ratings_driver = np.mean(df['Driver Ratings'])

print(f"Driver Average Rating: {avg_ratings_driver:.2f} / 5 stars")
```

```python
# Calculate Customer Average Ratings
avg_ratings_customer = np.mean(df['Customer Rating'])

print(f"Customer Average Rating: {avg_ratings_customer:.2f} / 5 stars")
```
